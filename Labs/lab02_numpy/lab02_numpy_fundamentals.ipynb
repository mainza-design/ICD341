## Part B: Array creation methods

```python
a = np.zeros(10)
b = np.ones(10)
c = np.arange(1, 11)
d = np.linspace(0, 1, 11)

print(a)
print(b)
print(c)
print(d)
```

**Questions**
1. `np.zeros()` creates an array filled with 0s.
2. `np.ones()` creates an array filled with 1s.
3. `arange(start, stop, step)` is defined by a **step size** and excludes the stop value. `linspace(start, stop, num)` is defined by the **number of points** and includes the stop value by default.
4. Statistical applications:
   - `zeros`: initialising an accumulator or results array (e.g. storing simulation results).
   - `ones`: the intercept column of a regression design matrix, or equal weights.
   - `arange`: observation indices or time points.
   - `linspace`: a grid of x-values for plotting a density or CDF.

## Part C: Indexing and slicing

```python
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

print(marks[0])      # first mark
print(marks[-1])     # last mark
print(marks[:3])     # first three marks
print(marks[2:6])    # positions 3 to 6
print(marks[::2])    # every second mark
```

## Part D: Statistical calculations

```python
print("Mean:  ", np.mean(marks))
print("Median:", np.median(marks))
print("Std:   ", np.std(marks))
print("Min:   ", np.min(marks))
print("Max:   ", np.max(marks))
print("Sum:   ", np.sum(marks))
```

**Task 3: Interpretation**
- The **mean (70.75)** is the average mark obtained by the students.
- The **median (70.0)** is the middle mark: half the students scored below it and half above.
- The **standard deviation (about 14.92)** measures the typical spread of marks around the mean (population formula, `ddof=0`).
- The **minimum (45)** and **maximum (91)** are the lowest and highest marks, a range of 46.
- The **sum (566)** is the total of all marks.

The mean and median are close, which suggests the distribution is roughly symmetric.

## Part E: Two-dimensional arrays

```python
data = np.array([
    [45, 67, 82],
    [56, 91, 73],
    [64, 88, 76]
])

print(data.shape)
print(data.ndim)
```

```python
print(data[0])       # 1. first row
print(data[:, 1])    # 2. second column
print(data[1, 1])    # 3. the value 91
print(data[:2])      # 4. first two rows
print(data[:, 1:])   # 5. last two columns
```

## Part F: Statistical application
Columns: Mathematics | Statistics | Programming

```python
print("Mean per subject:", np.mean(data, axis=0))
print("Mean per student:", np.mean(data, axis=1))
```

**`axis=0` vs `axis=1`**
- `axis=0` runs **down the rows**, collapsing them, so you get **one result per column** (here, per subject).
- `axis=1` runs **across the columns**, collapsing them, so you get **one result per row** (here, per student).

## Lab 2 Challenge

```python
import matplotlib.pyplot as plt

x = np.random.normal(loc=50, scale=10, size=1000)

print("Mean:  ", np.mean(x))
print("Median:", np.median(x))
print("Std:   ", np.std(x))
print("Min:   ", np.min(x))
print("Max:   ", np.max(x))

plt.hist(x, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.show()
```

**Observation:** the mean and median are both close to 50, the standard deviation is close to 10, and the histogram is bell-shaped and centred near 50, as expected for `Normal(50, 10²)`. The values differ slightly each run because no seed is set.
