## Task 3: `sys.version` vs `sys.executable`

- `sys.version` is a **string describing the Python interpreter that is running**: its version number, build date and compiler.
- `sys.executable` is the **file path of the interpreter binary** that is running your code (for example `.../envs/icd341/python.exe`). It tells you *which* Python, and therefore which environment, is in use.

Together they let you confirm you are using Python 3.12 from the `icd341` environment.

1. **What is a Python environment?** An isolated installation of Python plus a set of packages, kept separate from other installations on the machine.
2. **Why a separate environment for a course?** It avoids version conflicts with other projects and makes the setup reproducible and easy to rebuild or share.

## Lab 1 short-answer questions
1. **Python environment:** an isolated set of a Python interpreter and installed packages.
2. **Two advantages of Conda environments:** (i) isolation, so different projects can use different package versions without clashing; (ii) reproducibility, because environments can be exported and recreated.

7. **`git add` vs `git commit`:** `git add` stages changes (chooses what goes into the next snapshot); `git commit` records the staged changes as a permanent snapshot with a message.
8. **Purpose of `.gitignore`:** it tells Git which files and folders not to track, such as caches, secrets and generated files.
9. **Why exclude `.ipynb_checkpoints`:** these are auto-generated, machine-specific backups that change constantly, add clutter and cause merge conflicts, and contain no original work.
10. **Why reproducible environments matter:** results depend on Python and package versions. A reproducible environment lets you or others rerun the analysis and obtain the same results, which is essential for scientific validity.

**Environment vs repository:** an environment holds the *software* that runs your code; a Git repository holds your *files and their history*.
